### Answers to the Review Questions

**NumPy (Days 4–6)**
* **Why is a NumPy array faster than a Python list for math?** NumPy arrays are stored in contiguous blocks of memory and operations are implemented in optimized C code, avoiding Python's type-checking overhead per element.
* **What does axis=0 do versus axis=1 in an aggregation like .sum()?** `axis=0` aggregates down the rows (collapsing the rows, thus operating column-wise), while `axis=1` aggregates across the columns (collapsing the columns, thus operating row-wise).
* **Explain broadcasting in one sentence. What does `np.array([[1],[2],[3]]) + np.array([10,20,30])` produce?** Broadcasting allows NumPy to perform operations on arrays of different shapes by virtually replicating the smaller array to match the larger one. The operation produces a 3x3 array:
`[[11, 21, 31],`
` [12, 22, 32],`
` [13, 23, 33]]`


```python
import numpy as np

# create a 3x3 array counting 1-9
arr = np.arange(1, 10).reshape(3, 3)
print("Array:\n", arr)

# extract middle column
middle_col = arr[:, 1]
print("\nMiddle Column:", middle_col)

# extract last row
last_row = arr[-1, :]
print("Last Row:", last_row)
```

```python
import numpy as np

# given a random array, replace every value below its mean with 0
np.random.seed(42)
rand_arr = np.random.randint(1, 100, size=(4, 4))
print("Original Array:\n", rand_arr)
print("Mean:", rand_arr.mean())

rand_arr[rand_arr < rand_arr.mean()] = 0
print("\nModified Array:\n", rand_arr)
```

**Calculus & gradient descent (Days 17–18)**
* **What does a derivative tell you? What is the slope at a function's minimum?** A derivative tells you the instantaneous rate of change (the slope) of a function at a specific point. At a function's minimum, the slope is 0.
* **Explain gradient descent using the "hill in the fog" metaphor.** Imagine you're blindfolded on a foggy hill and want to reach the bottom. You feel the slope with your feet and take a step in the steepest downward direction, repeating until you reach a flat area (the minimum).
* **What does the learning rate control, and what happens if it's far too large?** It controls the size of the steps taken in the direction of the gradient. If it's too large, you might overshoot the minimum and bounce back and forth, diverging instead of converging.
* **In one sentence, connect gradient descent to how a model "learns."** A model learns by iteratively adjusting its internal parameters (weights) using gradient descent to move step-by-step down the error surface, minimizing its mistakes.


### Statistics (Day 19) - Code Examples

**1. Why is the median more robust to outliers than the mean?**

```python
import numpy as np
data = [10, 12, 15, 14, 13]
data_with_outlier = [10, 12, 15, 14, 13, 1000]

print("Mean without outlier:", np.mean(data), " | Median without outlier:", np.median(data))
print("Mean with outlier:   ", np.mean(data_with_outlier), "| Median with outlier:   ", np.median(data_with_outlier))
# The mean is heavily pulled by the outlier, the median barely changes.
```

**2. If a dataset's mean is much larger than its median, what does that tell you about its shape?**

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate a right-skewed dataset
data_skewed = np.random.lognormal(mean=0, sigma=1, size=1000)
mean_val = np.mean(data_skewed)
median_val = np.median(data_skewed)

print(f"Mean: {mean_val:.2f}, Median: {median_val:.2f}")

plt.hist(data_skewed, bins=50, alpha=0.7)
plt.axvline(mean_val, color='r', linestyle='dashed', linewidth=2, label='Mean')
plt.axvline(median_val, color='g', linestyle='dashed', linewidth=2, label='Median')
plt.legend()
plt.title("Right-Skewed Distribution")
plt.show()
```

**3. What does standard deviation measure, in plain words?**

```python
import numpy as np

data1 = [50, 50, 50, 50, 50]  # No spread
data2 = [0, 25, 50, 75, 100]  # High spread

print("Data 1 STD:", np.std(data1), "(All values are the mean)")
print("Data 2 STD:", np.std(data2), "(Values are spread out)")
```

### Probability (Day 20) - Code Examples

**1. What is conditional probability, P(A|B), in plain words?**

```python
# Example: Rolling a 6-sided die. 
# A: roll is 6
# B: roll is even (2, 4, or 6)

even_rolls = [2, 4, 6]
is_six = [roll == 6 for roll in even_rolls]

# Probability of rolling a 6, GIVEN we know the roll was even
p_A_given_B = sum(is_six) / len(even_rolls)
print(f"P(Roll is 6 | Roll is even) = {p_A_given_B:.2f} (or 1/3)")
```

**2. The famous one: a disease affects 1% of people; a test is 99% accurate...**

```python
# Out of 10,000 people
n_people = 10000
disease_rate = 0.01
accuracy = 0.99

n_sick = int(n_people * disease_rate) # 100
n_healthy = n_people - n_sick # 9900

true_positives = int(n_sick * accuracy) # 99
false_positives = int(n_healthy * (1 - accuracy)) # 99

total_positives = true_positives + false_positives
chance_of_having_disease = true_positives / total_positives

print(f"Total positive tests: {total_positives}")
print(f"Actually sick among positives: {true_positives}")
print(f"Chance you actually have it if you test positive: {chance_of_having_disease:.2%}")
```

**3. Why can '99% accuracy' be a misleading metric when the thing you're detecting is rare?**

```python
from sklearn.metrics import accuracy_score, confusion_matrix

# Dataset has 99 normal transactions and 1 fraudulent one
y_true = [0]*99 + [1]

# A 'dumb' model predicts everything is normal (0)
y_pred_dumb = [0]*100

print("Accuracy:", accuracy_score(y_true, y_pred_dumb))
print("Confusion Matrix:\n", confusion_matrix(y_true, y_pred_dumb))
# High accuracy (99%), but it completely failed to find the 1 fraud case!
```

### Synthesis (Day 22) - Code Examples

**1. Trace the thread: how does a dot product relate to gradient descent relate to model.fit()?**

```python
import numpy as np

# 1. Dot product: How a linear model makes predictions internally
x = np.array([1, 2])     # input features
w = np.array([0.5, 0.5]) # internal weights
prediction = np.dot(x, w) 
print("Prediction via dot product:", prediction)

# 2. Gradient Descent: Updating weights based on error
target = 2.0
error = prediction - target
learning_rate = 0.1
# Update rule (simplified gradient descent step): w = w - learning_rate * error * x
w_new = w - learning_rate * error * x
print("Updated weights after 1 step of gradient descent:", w_new)

# 3. model.fit(): Automates this process over many iterations (epochs) to find the perfect weights!
```
