# Calculus to Code: Connecting Derivatives to Machine Learning

Today we bridge the gap between abstract math and actionable code. We will visually explore slopes, compute derivatives numerically, and understand the intuition behind Gradient Descent.

### 1. Plotting a simple function

Let's plot $y = x^2$. By eye, notice where the slope is steep and where it is flat. The flat point at the bottom is the minimum!

```python
import numpy as np
import matplotlib.pyplot as plt

# Create an array of x values from -5 to 5
x = np.linspace(-5, 5, 100)

# Calculate the corresponding y values (y = x^2)
y = x**2

# Plot the curve
plt.figure(figsize=(8, 5))
plt.plot(x, y, label='y = x^2', color='blue', linewidth=2)

# Highlight a few points
plt.scatter([-3, 0, 3], [9, 0, 9], color='red', zorder=5)
plt.annotate('Steep negative slope', (-3, 9), textcoords="offset points", xytext=(-80,10), ha='center')
plt.annotate('Steep positive slope', (3, 9), textcoords="offset points", xytext=(50,10), ha='center')
plt.annotate('Flat slope (Minimum)', (0, 0), textcoords="offset points", xytext=(0,20), ha='center')

plt.title('Plot of y = x^2')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.show()
```

### 2. Approximating the Derivative Numerically

The formal definition of a derivative is `(f(x + h) - f(x)) / h` as `h` approaches zero. We can write a simple Python function to approximate this by choosing a very tiny `h`.

```python
def f(x):
    return x**2

def approximate_derivative(f, x, h=0.0001):
    # Calculate the slope over a tiny interval h
    return (f(x + h) - f(x)) / h

# Let's test it at three critical points on our x^2 curve:
left_point = -3.0
bottom_point = 0.0
right_point = 3.0

slope_left = approximate_derivative(f, left_point)
slope_bottom = approximate_derivative(f, bottom_point)
slope_right = approximate_derivative(f, right_point)

print(f"Slope at x={left_point}: {slope_left:.2f} (Negative slope!)")
print(f"Slope at x={bottom_point}: {slope_bottom:.2f} (Zero slope!)")
print(f"Slope at x={right_point}: {slope_right:.2f} (Positive slope!)")
```

### 3. The Connection to Machine Learning

This is the whole point! 

A Machine Learning model has an **error function** (or loss function) which measures how wrong its predictions are. Training a model simply means finding the bottom of that error curve — the **minimum**.

The derivative (slope) tells the model which direction reduces the error. The model looks at the slope, takes a small step in the opposite direction (downhill), and recalculates. Repeat this thousands of times, and the model "learns."

#### The Metaphor: Gradient Descent

> Sit with this metaphor: You are standing on a hill in dense fog, wanting to reach the valley. You can't see the bottom, but you can feel the slope under your feet. By taking steps in the direction that slopes downward the steepest, you will eventually reach the valley. 
>
> The derivative is your ability to **feel the slope**. This process is called **Gradient Descent**, and it is the absolute engine of deep learning.
