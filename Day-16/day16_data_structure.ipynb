Connect to ML: a neural network layer is literally inputs @ weights + bias. One matrix multiply computes every neuron's output at once. The thing you're learning today is the forward pass of a neural network — sit with that.

### 1. Matrix Multiplication & Shapes
First, let's create the matrices, multiply them, and check the resulting shape.


```python
import numpy as np

# Create a 2x3 matrix (2 rows, 3 columns)
matrix_a = np.array([[1, 2, 3], 
                     [4, 5, 6]])

# Create a 3x2 matrix (3 rows, 2 columns)
matrix_b = np.array([[7, 8], 
                     [9, 10], 
                     [11, 12]])

# Multiply them using the @ operator
result = matrix_a @ matrix_b

print("Matrix A shape:", matrix_a.shape)
print("Matrix B shape:", matrix_b.shape)
print("\nResult of A @ B:\n", result)
print("Result shape:", result.shape)
```

### 2. The Dreaded Shape Error (Incompatible Shapes)
Let's intentionally try to multiply two matrices where the inner dimensions don't match.


```python
# Create a 2x4 matrix
matrix_c = np.array([[1, 2, 3, 4], 
                     [5, 6, 7, 8]])

# Try multiplying the (2x3) Matrix A by the (2x4) Matrix C
try:
    bad_result = matrix_a @ matrix_c
except ValueError as e:
    print("Caught an error!")
    print(e)
```

### 3. Hand-Computing One Entry
Let's compute the top-left entry of `matrix_a @ matrix_b` by hand to prove the math works.


```python
# To get the top-left entry (index 0,0), we take:
# Row 0 of Matrix A: [1, 2, 3]
# Col 0 of Matrix B: [7, 9, 11]

# Manual dot product: (1 * 7) + (2 * 9) + (3 * 11)
manual_entry = (1 * 7) + (2 * 9) + (3 * 11)

print("Manual computation of top-left entry:", manual_entry)
print("NumPy computation of top-left entry:", result[0, 0])

assert manual_entry == result[0, 0], "They match!" 
```

### 4. Transposing (.T)
Let's fix our shape mismatch from Step 2 by transposing `matrix_c`.


```python
print("Original Matrix C shape:", matrix_c.shape)

# Transpose Matrix C (flips rows and columns)
matrix_c_transposed = matrix_c.T

print("Transposed Matrix C shape:", matrix_c_transposed.shape)

# Now we CAN multiply Matrix A (2x3) and Matrix C_transposed (4x2)! Wait... no we can't!
# (2x3) @ (4x2) still fails. 

# BUT we can multiply C.T (4x2) @ A (2x3)
result_transposed = matrix_c_transposed @ matrix_a

print("\nResult of C.T @ A shape:", result_transposed.shape)
```

### 5. The Connection to Machine Learning

In a neural network, everything is matrix multiplication:
*   **`X` (Inputs):** A matrix of your data (e.g., a batch of 32 images, each with 784 pixels -> `32 x 784` matrix).
*   **`W` (Weights):** A matrix representing the connections to the next layer of neurons (e.g., 784 inputs mapping to 128 hidden neurons -> `784 x 128` matrix).

When you compute `Outputs = X @ W`, you are taking the dot product of your input features with every single neuron's weights *simultaneously*.
*   `(32 x 784) @ (784 x 128) = (32 x 128)` 
*   Your output is instantly a perfectly shaped matrix representing the activation of 128 neurons for all 32 images in the batch. 

That is the **Forward Pass**. GPUs are essentially super-calculators built specifically to do these `m × n @ n × p` operations millions of times per second.
