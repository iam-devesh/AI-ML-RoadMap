In NumPy, create two vectors. Add them, scale one by 3, and compute their dot product with np.dot().
Compute a dot product by hand (multiply matching elements, sum them), then confirm NumPy agrees. Seeing the formula match the code locks it in.
Compute the length (magnitude) of a vector: np.sqrt(np.sum(v**2)) or np.linalg.norm(v).
Two vectors pointing the same way have a large positive dot product; perpendicular ones have a dot product of ~0. Create examples and verify this yourself.

```python
import numpy as np

# 1. Create two vectors. Add them, scale one by 3, and compute their dot product with np.dot().
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])

print("Vector 1:", v1)
print("Vector 2:", v2)

# Add them
v_add = v1 + v2
print("\nAddition (v1 + v2):", v_add)

# Scale one by 3
v1_scaled = v1 * 3
print("Scaled Vector 1 (v1 * 3):", v1_scaled)

# Compute their dot product with np.dot()
dot_np = np.dot(v1, v2)
print("Dot product (np.dot):", dot_np)

# 2. Compute a dot product by hand (multiply matching elements, sum them), then confirm NumPy agrees.
dot_hand = sum(x * y for x, y in zip(v1, v2))
print("\nDot product (by hand):", dot_hand)
print("Does NumPy agree?", dot_np == dot_hand)

# 3. Compute the length (magnitude) of a vector
mag_hand = np.sqrt(np.sum(v1**2))
mag_np = np.linalg.norm(v1)
print("\nMagnitude of v1 (by formula):", mag_hand)
print("Magnitude of v1 (np.linalg.norm):", mag_np)

# 4. Two vectors pointing the same way have a large positive dot product; perpendicular ones have a dot product of ~0. 

# Vectors pointing the same way (parallel)
v_same1 = np.array([1, 0])
v_same2 = np.array([3, 0])
dot_same = np.dot(v_same1, v_same2)
print("\nVectors pointing the same way:")
print("v_same1 =", v_same1, ", v_same2 =", v_same2)
print("Dot product:", dot_same)

# Perpendicular vectors (orthogonal)
v_perp1 = np.array([1, 0])
v_perp2 = np.array([0, 1])
dot_perp = np.dot(v_perp1, v_perp2)
print("\nPerpendicular vectors:")
print("v_perp1 =", v_perp1, ", v_perp2 =", v_perp2)
print("Dot product:", dot_perp)
```
