Take a simple 2-input function like f(x, y) = x² + y² (a bowl shape — minimum at the origin). Its gradient is [2x, 2y].
Write a tiny gradient descent loop in NumPy: start at a random point, repeatedly do point = point - learning_rate * gradient, and print the point each step. Watch it crawl toward (0, 0).
Experiment with the learning rate: try a tiny one (slow crawl), a good one (smooth descent), and a huge one (watch it diverge/explode). Seeing it break is genuinely instructive.
Connect to ML — this is the payoff: this exact loop, scaled to millions of parameters, is how every neural network trains. The error function replaces x² + y², the parameters replace x, y, but the algorithm is identical. You just wrote, in miniature, the engine of deep learning.

### 1. The Gradient Descent Engine

Here we define our function to compute the gradient of $f(x, y) = x^2 + y^2$, which is mathematically $[2x, 2y]$. 
We then loop for a set number of steps, updating our position based on the gradient and the learning rate.

```python
import numpy as np

def compute_gradient(point):
    # The derivative of x^2 is 2x, and y^2 is 2y.
    # So the gradient vector is [2x, 2y]
    x, y = point
    return np.array([2 * x, 2 * y])

def run_gradient_descent(start_point, learning_rate, steps=10):
    point = np.array(start_point, dtype=float)
    print(f"--- Starting at {point} with Learning Rate = {learning_rate} ---")
    
    for step in range(1, steps + 1):
        gradient = compute_gradient(point)
        
        # The core Gradient Descent update rule:
        # Move in the NEGATIVE direction of the gradient
        point = point - (learning_rate * gradient)
        
        # Print every step to watch it move
        print(f"Step {step:2}: Point = [{point[0]:.4f}, {point[1]:.4f}]")
        
    print("Final position:", point, "\n")
```

### 2. Experimenting with Learning Rates

Let's see what happens when we use different learning rates. Notice how the 'Good' learning rate smoothly converges to `[0, 0]`.

```python
start = [10.0, 10.0]

# 1. A GOOD learning rate (smooth, fast convergence)
run_gradient_descent(start, learning_rate=0.1, steps=15)

# 2. A TINY learning rate (painfully slow, barely moves)
run_gradient_descent(start, learning_rate=0.01, steps=15)

# 3. A HUGE learning rate (overshoots and explodes!)
run_gradient_descent(start, learning_rate=1.1, steps=15)
```

### 3. The Connection to Machine Learning 🧠

What you just wrote above is the exact mechanism that powers ChatGPT, self-driving cars, and image generators.

*   **The Function:** Instead of a simple bowl $x^2 + y^2$, deep learning uses an **Error Function** (Loss). The goal is exactly the same: reach the absolute bottom (0 error).
*   **The Point:** Instead of a 2D coordinate `[x, y]`, neural networks have points in millions or billions of dimensions. These are the **weights** of the network.
*   **The Algorithm:** The update rule `point = point - learning_rate * gradient` is executed identically. PyTorch computes the massive gradients for you (using backpropagation), and an optimizer takes the step.

When a neural network is "training", it is literally just looping the code block you wrote above millions of times, inching its way to the bottom of the error bowl.
